```python
import pandas as pd
```

```python
import json

with open('report.json', 'r') as file:
    raw_data = json.load(file)

normalized_data = pd.json_normalize(raw_data)
```

```python
import re
data=raw_data['sections']
usedcolumnsInreport=[]
for dt in data:
    #print(dt['displayName'])
    #print(dt['name'])
    for e in dt['visualContainers']:
        jsondata=json.loads(e['config'])
       
        if 'projections' in jsondata['singleVisual']:
                      
            for key, value in jsondata['singleVisual']['projections'].items():
                
                for key1, value1 in value[0].items():
                    if 'queryRef' in  key1:
                        pattern = r"([a-zA-Z0-9_]+)\.([a-zA-Z0-9_]+)"  #to get table column names from the DAX expressions
                        tablecolumn=re.findall(pattern, value1)
                        for table, column in tablecolumn:
                            obj=[dt['displayName'],dt['name'],jsondata['name'],jsondata['singleVisual']['visualType'],table,column,value1]
                            usedcolumnsInreport.append(obj)

finaldataframe = pd.DataFrame(usedcolumnsInreport, columns=['Page Name','Page Id','Visual Id','Visual Type','Table', 'Column','DAX Expression'])
finaldataframe.to_csv('used_columns_in_report.csv', index=False)                 

                
                
                
           
            
            
            
      
```
